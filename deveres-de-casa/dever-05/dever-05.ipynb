## Apresentar o cálculo de complexidade para:
#### Algoritmo de ordenação do Merge Sort
$$T(n) = 2T\left(\frac{n}{2}\right) + f(n)$$
$2T(n/2)$: Representa as duas chamadas recursivas para as metades do array.

$f(n) = \Theta(n)$: Representa o custo da função Merge (intercalação), que percorre todos os elementos para juntá-los em ordem.
$T(n) = aT(n/b) + f(n)$.No caso do Merge:
$a = 2$ (número de subproblemas)$ b = 2$ (fator de divisão do tamanho do problema)$f(n) = n^1$
$$\log_2 2 = 1$$

Caso 2 do Teorema Mestre:
$$T(n) = \Theta(n^{\log_b a} \log n) = \mathbf{\Theta(n \log n)}$$

#### Multiplicação de matrizes
```python
for i from 1 to n:          # Loop 1
    for j from 1 to n:      # Loop 2
        C[i][j] = 0
        for k from 1 to n:  # Loop 3
            C[i][j] += A[i][k] * B[k][j]
```
Análise de Complexidade:
- Loop Externo ($i$): Executa $n$ vezes.
- Loop Intermediário ($j$): Executa $n$ vezes para cada iteração de $i$.
- Loop Interno ($k$): Executa $n$ vezes para cada combinação de $i$ e $j$.

$$T(n) = n \cdot n \cdot n = n^3$$

#### Recorrências:

a) $T(n) = 2T(n/4) + \sqrt{n}$

Identificação: $a = 2, b = 4, f(n) = \sqrt{n} = n^{1/2}$.

Cálculo do termo crítico: $\log_b a = \log_4 2 = 0.5$ (pois $4^{0.5} = 2$).

Comparação: O termo crítico é $n^{0.5}$, que é exatamente igual a $f(n) = n^{1/2}$.

Caso do Teorema Mestre: Caso 2 (quando $f(n) = \Theta(n^{\log_b a})$).

Resultado: $T(n) = \Theta(n^{\log_b a} \log n) = \mathbf{\Theta(\sqrt{n} \log n)}$.

b) $T(n) = 2T(n/4) + n$

Identificação: $a = 2, b = 4, f(n) = n^1$.

Cálculo do termo crítico: $\log_4 2 = 0.5$.

Comparação: Comparamos $n^{0.5}$ com $f(n) = n^1$. Como $1 > 0.5$, a função $f(n)$ cresce mais rápido.

Caso do Teorema Mestre: Caso 3 (quando $f(n) = \Omega(n^{\log_b a + \epsilon})$).

Resultado: A complexidade é dominada por $f(n)$.Resposta: $T(n) = \mathbf{\Theta(n)}$.

c) $T(n) = 16T(n/4) + n^2$

Identificação: $a = 16, b = 4, f(n) = n^2$.

Cálculo do termo crítico: $\log_4 16 = 2$ (pois $4^2 = 16$).

Comparação: O termo crítico é $n^2$, que é exatamente igual a $f(n) = n^2$.

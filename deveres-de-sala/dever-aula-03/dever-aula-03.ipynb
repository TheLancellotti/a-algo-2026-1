```python
def encontrar_substrings(s, inicio=0, fim=0):
    # PASSO 1:
    # Recursão para quando o índice de inicio atinge o final da string.
    if inicio == len(s):
        return

    # PASSO 2:
    # O índice fim ainda não chegou ao final, imprimim a substring atual e avança o fim.
    elif fim < len(s):
        print(s[inicio : fim + 1])
        encontrar_substrings(s, inicio, fim + 1)

    # PASSO 3: 
    # Quando o fim atinge o limite, reinicia o processo a partir do próximo caractere e reseta o fim.
    else:
        encontrar_substrings(s, inicio + 1, inicio + 1)

    # PASSO 4:
    # A função retorna None após esgotar as chamadas recursivas.

string = "abcd"
print(f"Substrings de '{string}':")
encontrar_substrings(string)
```

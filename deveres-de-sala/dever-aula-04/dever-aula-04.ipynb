```python
def recorrencia(n, r):
    if n == 0:
        return r
    elif n == 1:
        r = n - 1 + n + 1 + r
        return r
    else:
        r = n - 1 + n + 1 + r
        return recorrencia(n-1, r)
```

```python
n = int(input("Digite o valor de n: "))
print(recorrencia(n, 0))
```

```python
def recorrencia2(n, r):
    if n == 1:
        return 1 + r
    else:
        r = n - 1 + 3 * n + 2 + r
        return recorrencia2(n-1, r)
```

```python
n = int(input("Digite o valor de n: "))
print(recorrencia2(n, 0))
```
